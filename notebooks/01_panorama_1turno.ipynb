Dados lidos direto da API oficial de Resultados do TSE (sem download).

```python
from tse.resultados_api import ResultadosTSE
import pandas as pd
api = ResultadosTSE()
pd.DataFrame([api.totais(1)]).T
```

```python
api.candidatos(1)[['numero','nome_urna','partido','votos','pct_validos','situacao']]
```

```python
ufs = api.totais_ufs(1)
ufs[['uf','eleitorado','pct_comparecimento','pct_brancos','pct_nulos']].sort_values('pct_comparecimento')
```

```python
# Presidente por UF: votos de cada candidato
cand_uf = pd.concat([api.candidatos(1, uf) for uf in ufs['uf']])
cand_uf.pivot_table(index='uf', columns='nome_urna', values='pct_validos').round(2)
```
